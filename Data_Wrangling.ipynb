# **Combining data using merge and join**

```python
import pandas as pd
students=pd.DataFrame({
    'student_id':[101,102,103,104],
    'Name':['Aman','Priya','Rahul','Shagun'],
    'Dept':['CSE','CSE','DS','ECE']
})
marks=pd.DataFrame({
    'student_id':[101,102,104,105],
    'Marks':[87,72,91,86]
})
```

```python
print("Student table:")
print(students)
print("\nmarks table:")
print(marks)
```

```python
inner=pd.merge(students,marks,on='student_id',how='inner')
print("\n Inner join")
print(inner)
```

```python
left=pd.merge(students,marks,on='student_id',how='left')
print("\n Left join")
print(left)
```

```python
outer=pd.merge(students,marks,on='student_id',how='outer')
print("\n Outer join")
print(outer)
```

# **Column names differ**

```python
student_info=students.rename(columns={'student_id':'id'})
print("\n Student information")
print(student_info)
combined=pd.merge(
    student_info,marks,left_on='id',right_on='student_id',how='inner'
)
print("\n Combined data:")
print(combined)
```

# **Join using an index**

```python
students_indexed=students.set_index('student_id')
marks_indexed=marks.set_index('student_id')
joined=students_indexed.join(marks_indexed,how='left')
print("\nStudents(indexed)")
print(students_indexed)
print("\n Marks(indexed)")
print(marks_indexed)
print("\n Joined")
print(joined)
```
